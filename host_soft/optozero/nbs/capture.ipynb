```python
%matplotlib notebook
import cv2
import matplotlib.pyplot as plt
from IPython.display import display
import numpy as np
```

```python
cap = cv2.VideoCapture('http://localhost:8082/')
ret, img = cap.read()
cap.release()

ret, circles = cv2.findCirclesGrid(img, (6,6))
if ret:
    cv2.drawChessboardCorners(img, (6, 6), circles, ret)
    #display(circles)
    x = circles[:,0][:,0]
    y = circles[:,0][:,1]
    display([np.average(x), np.average(y)])
    
plt.imshow(img)
```
